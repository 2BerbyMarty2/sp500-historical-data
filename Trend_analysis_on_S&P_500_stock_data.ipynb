**The "Best" Stock**
* If you bought 1 share of *every* stock on the first day of the dataset and sold it on the last day, which stock would have given you the highest percentage profit?

```python
df = df.sort_values(by = 'date')

start_prices = df.groupby('Name')['open'].first()
end_prices = df.groupby('Name')['close'].last()

profit_percentages = ((end_prices - start_prices) / start_prices) * 100

best_stock = profit_percentages.idxmax()

print(f"The best stock to buy and hold is {best_stock} with a profit percentage of {profit_percentages[best_stock]:.2f}%")
```

```python
# 1. Sort by date first so 'first' and 'last' are actually chronological
df = df.sort_values('date')

# 2. Group by ticker and grab the first 'open' and last 'close'
start_prices = df.groupby('Name')['open'].first()
end_prices = df.groupby('Name')['close'].last()

# 3. Calculate profit directly (Pandas handles the math for all rows at once)
# Formula: (Sell Price - Buy Price) / Buy Price
profits = (end_prices - start_prices) / start_prices

# 4. Sort and print the top 5
print(profits.sort_values(ascending=False).head())
```

```python
df['date'] = pd.to_datetime(df['date'])

# Filter for the winner
winner_df = df[df['Name'] == best_stock]

# Plot
plt.figure(figsize=(10, 5))
plt.plot(winner_df['date'], winner_df['close'])
plt.title(f"Price History of {best_stock}")
plt.show()
```
